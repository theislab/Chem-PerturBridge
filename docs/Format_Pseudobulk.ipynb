**NB!** The standardization/validation is performed by using `bionty==1.5.0` and `wetlab==1.2.0` for **Tahoe**, and `bionty==1.0.0` and `wetlab==1.0.0` for **Sciplex**. The config files with the default database versions (the latest in the config) used for the standardization are located: [bionty==1.0.0](https://github.com/laminlabs/bionty/blob/1.0.0/bionty/base/sources.yaml), [bionty==1.5.0](https://github.com/laminlabs/bionty/blob/1.5.0/bionty/base/sources.yaml))

`plate <category>` - is a column which indicates the experimental plate, which was used to perform perturbation; it is written in the format `platei`. **NB!** Make sure that cells in this column do only contain letters and digits (or `.` in the worst case).

`well <category>` - is a column which indicates the id of the well on the experimental plate, where the perturbation was performed; the format might vary, but we tried to stick to letter-number annotation, e.g. `A9`.

`cell_type <category>` - is a column indicating a cell type used in the experiments, cellosaurus ids are used to identify the cancer cell line. No the specific version of the **Cellosaurus** database is used.

`perturbagen <category>` - indicates the name of perturbagen. They are taken from the initial datasets and standardized by the lamin lab using `chebi` with the version `2024-07-27` for both **Sci-Plex** and **Tahoe**.

`pert_type <category>` - indicates the type of perturbagen (e.g. `compound`, `genetic`, `biologic`, `physical`: [docs for `wetlab==1.0.0`](https://github.com/laminlabs/wetlab/blob/1.0.0/docs/guide/pert-curator.ipynb)). The types are standardized by lamin lab.

`is_control <bool>` - indicates if the perturbation is control or not. (Currently, if the perturbagen is `DMSO` - then perturbation is control)

`pert_dose_uM <float64>` - the information on the concentration, which is converted to standard units (uM) by the python package `magnitude==1.1.2`.

`pert_time_h <float64>` - the information on the perturbation duration, which is converted to standard units (hours) by the python package `magnitude==1.1.2`.

`suspension_type <category>` - what kind of biological material was dissociated into single cells or nuclei, manually annotated by the lamin lab.

`tissue <category>` - the information on the tissue related to the cell line, manually annotated by the lamin lab; validated with (`uberon 2024-08-07`). 

`tissue_type <category>` - the information on the tissue type related to the cell line (e.g. `cell culture`), manually annotated by the lamin lab. There are pipelines for cell line annotations (`suspension_type`, `tissue`, `tissue_type`, `assay`) : [**Sci-Plex**](https://lamin.ai/laminlabs/pertdata/transform/qNq33VZmhe4t0000/fGwen5GH3Zixpehg8iaN), [**Tahoe**](https://lamin.ai/laminlabs/pertdata/transform/MQDTES9N9RCl0000).

`disease <category>` - indicates the name of disease (the type of tumor) related to the cell line, standardized by the lamin lab. For both **Sci-Plex** and **Tahoe**: `mondo` with `2024-08-06` version, `doid` with `2024-05-29` version.

`library <category>` - indicates the ID of the library which is used to perform sequencing; in some cases the info might not exist.

`stimulation <category>` - information on the stimuli, additional conditions which cells are exposed to during the perturbation; in some cases the info might not exist.

`guide <category>` - relevant to the genetic perturbations (e.g. for CRISPR-based ones it determines a guide RNA directs the CRISPR system); in some cases the info might not exist.

`dataset <category>` - a name of the dataset, specified manually. There is no specific format.

`assay <category>` - the info on the experimental method used to capture the information for the construction the dataset, manually annotated by the lamiin lab, both datasets were validated with `efo 3.70.0`.

`development_stage <category>` - the stage of the development of the tissue from the patient, might include age (e.g. `51-year-old stage`, or `unknown`). It was standardized and validated by the lamin lab (the db for validation for both datasets: `hsapdv 2024-05-28`).

`organism <category>` - a type of an organism, which cell line is used for the perturbation experiment, we consider `human` cells. It was standardized and validated by the lamin lab (the db for validation for both datasets: `ensembl release 112`).

`sex <category>` - a gender of an organism, which cell line is used for the perturbation experiment. It is specified manually by the lamin lab, validated with `pato 2024-03-28`.

`self_reported_ethnicity <category>` - an ethnicity of a human, which cell line is used for the perturbation experiment. It is specified manually by the lamin lab, validated with `hancestro 3.0` for **Sci-Plex** and `hancestro 2025-04-01` for **Tahoe**. There are pipelines for patient annotations (`development_stage`, `organism`, `sex`, `self_reported_ethnicity`) : [Sci-Plex](https://lamin.ai/laminlabs/pertdata/transform/qNq33VZmhe4t0000/fGwen5GH3Zixpehg8iaN), [Tahoe](https://lamin.ai/laminlabs/pertdata/transform/dtQBk1ocZHdN0001/6I5be4yXJPdDSyeh).

`pubchem_cid <category>` - the **PubChem** CIDs retrieved by using the python package `pubchempy==1.0.5`.

`psbulk_cells <int64>` - the number of cells which are included in pseudobulk (obtained by using `.pp.pseudobulk` function from the `decoupler==2.1.0`).

`psbulk_counts <int64>` - the number of counts which are included in pseudobulk (obtained by using `.pp.pseudobulk` function from the `decoupler==2.1.0`).

### index:

The index `sample_id <object>` for pseudobulk is constructed as `plate` + '\_' + `well` + '\_' + `perturbagen` + '\_' + `cell_type`

## .var

### columns:

.var includes one column: `symbol`

`symbol <category>` - mainly the human gene symbol from HGNC database, but also other ids from the original datasets might be used (e.g. `AC015801.1`).

```python
adata.var.columns
```

```python
adata.var.dtypes
```

```python
adata.var
```

### index:

The index `ensembl_id <object>` for **.var** is constructed as ID from **Ensemble** database. For [Sci-plex](https://lamin.ai/laminlabs/pertdata/transform/qNq33VZmhe4t0000) validation the default (`release-112`) version is used but as it is [found](https://github.com/theislab/op3_v2/blob/slurm_scripts_notebooks/draft_notebooks/a_trial_to_find_release_of_ensembl_for_sciplex.ipynb) the probable ensembl releases are `90` or `91`; for [Tahoe](https://lamin.ai/laminlabs/pertdata/transform/OlAaeZKlSyri0000/0LF29cNOvRPwM3Is) the `release-109` version is used.

The `.layers['psbulk_props']` object is a sparse matrix which contain information about the proportion of cells that have a non-zero value for a given feature (obtained by using `.pp.pseudobulk` function from the `decoupler==2.1.0`). I do not think it is 100% necessary information, so it could be ommitted, if it is impossible to receive.
