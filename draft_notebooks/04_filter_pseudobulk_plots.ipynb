```python
import scanpy as sc
import anndata as ad
```

```python
import matplotlib.pyplot as plt
```

```python
adata_sci = ad.read_h5ad('./data/sciplex/pseudobulk/full/srivatsan20_sciplex3.h5ad')
```

```python
adata_tahoe = ad.read_h5ad('./data/tahoe/pseudobulk/full/tahoe.h5ad')
```

```python
threshold = [0, 10, 25, 50, 100, 250]
```

```python
n_cells_s = []
n_cells_t = []
for t in threshold:
    n_cells_s.append((adata_sci.obs['psbulk_cells'] > t).sum())
    n_cells_t.append((adata_tahoe.obs['psbulk_cells'] > t).sum())
```

```python
n_cells_s
```

```python
n_cells_t
```

```python
plt.plot(threshold, n_cells_s, '-o')
plt.title('sciplex')
```

```python
plt.plot(threshold, n_cells_t, '-o')
plt.title('tahoe')
```
