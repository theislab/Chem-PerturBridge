### Download phase1 data

```python
#Download usual l1000 phase1
l1000_1 = ad.read_h5ad('/lustre/groups/ml01/workspace/olga.novitskaia/data_updated/l1000_phase1/pseudobulk_processed/sep_rep/l1000_phase1_level3_deg_ready_landmark_processed.h5ad')
```

```python
#Download level5 l1000 phase1
l1000_lvl5_1 = ad.read_h5ad('/lustre/groups/ml01/workspace/olga.novitskaia/data_updated/l1000_phase1/level5/level5_phase1_not_filtered.h5ad')
```

```python
#Show usual l1000 phase1 cell_type - perturbagen_name - pubchem_cid - pert_type

l1000_1.obs[['cell_type', 'perturbagen_name', 'pubchem_cid', 'pert_type']].drop_duplicates()
```

```python
#Show level5 l1000 phase1 cell_type - perturbagen_name - pubchem_cid - pert_type

l1000_lvl5_1.obs[['cell_type', 'perturbagen_name', 'pubchem_cid', 'pert_type']].drop_duplicates()
```

### Organize sets of (cell type - pubchem_cid) pairs

```python
# Organizing sets to overlap:

set_sci = set(sci.obs[['cell_type', 'pubchem_cid']].drop_duplicates().itertuples(index=False, name=None))

set_l1000_1 = set(l1000_1.obs[['cell_type', 'pubchem_cid']].drop_duplicates().itertuples(index=False, name=None))

set_l1000_lvl5_1 = set(l1000_lvl5_1.obs[l1000_lvl5_1.obs['pert_type'] == 'compound'][['cell_type', 'pubchem_cid']].drop_duplicates().itertuples(index=False, name=None))
```

### Overlap

```python
# Overlap Sciplex - l1000 phase1

len(set_sci.intersection(set_l1000_1))
```

```python
# Overlap Sciplex - lvl5 l1000 phase1

len(set_sci.intersection(set_l1000_lvl5_1))
```

```python
# Overlap  (Sciplex - l1000 phase1) - (Sciplex - lvl5 l1000 phase1)

pd.DataFrame(set_sci.intersection(set_l1000_1).intersection(set_sci.intersection(set_l1000_lvl5_1)), columns=['cell_type', 'pubchem_cid'])
```

```python
# Overlap  l1000 phase1 - lvl5 l1000 phase1

len(set_l1000_1.intersection(set_l1000_lvl5_1))
```

```python
# Differences  l1000 phase1 - lvl5 l1000 phase1

set_l1000_1 - set_l1000_lvl5_1
```

```python
# Differences  lvl5 l1000 phase1 - l1000 phase1

set_l1000_lvl5_1 - set_l1000_1
```

### Organize sets of (cell type - pubchem_cid - perturbagen_name) triplets

```python
set_l1000_1 = set(l1000_1.obs[['cell_type', 'pubchem_cid', 'perturbagen_name']].drop_duplicates().itertuples(index=False, name=None))

set_l1000_lvl5_1 = set(l1000_lvl5_1.obs[l1000_lvl5_1.obs['pert_type'] == 'compound'][['cell_type', 'pubchem_cid', 'perturbagen_name']].drop_duplicates().itertuples(index=False, name=None))
```

### Overlap

```python
len(set_l1000_1.intersection(set_l1000_lvl5_1))
```

```python
set_l1000_1 - set_l1000_lvl5_1
```

```python
set_l1000_lvl5_1 - set_l1000_1
```
