```python
import scanpy as sc
from tqdm import tqdm
import re
def remove_non_alphanumeric(s):
    return re.sub(r'[^a-zA-Z0-9]', '', s)
import numpy as np
import pandas as pd
import scanpy as sc
from tqdm import tqdm
import anndata

import matplotlib
import matplotlib.pyplot as plt
import seaborn as sns

# Some plotting aesthetics
%matplotlib inline

sns.set_style('ticks')
matplotlib.rcParams['figure.figsize'] =[4, 4]
matplotlib.rcParams['figure.dpi'] = 100
```

```python
def sum_adata(a0):
    a0n = anndata.AnnData(X=np.array(a0.X.sum(0)))
    a0n.obs = a0.obs[['sample', 'cell_name', 'bc1_well', 'drugname_drugconc']].iloc[:1]
    a0n.obs['n_cells'] = len(a0)
    a0n.var = a0.var.copy()
    return(a0n)
```

```python
for plate in range(1,15):
    adata = sc.read_h5ad(f'/lustre/groups/ml01/workspace/weixu.wang/100m/plate{plate}_filtered.h5ad')
    celllines = sorted(set(adata.obs.cell_name))
    wells = sorted(set(adata.obs.bc1_well))
    
    
    adatas0 = []
    for well in wells:
        for cellline in celllines:
            print(cellline)
            a0 = adata[(adata.obs.bc1_well==well)&(adata.obs.cell_name==cellline)].copy()
            if len(a0)>0:
                a0n = sum_adata(a0)
                adatas0.append(a0n)
    adata0 = anndata.concat(adatas0)
    adata0.write('../data/sub_adatas/bulked_sc/'+str(plate)+'.h5ad')
```
